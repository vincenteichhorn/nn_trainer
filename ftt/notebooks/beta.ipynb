```python

savings = 0.25
concentration = 2

alpha = savings * concentration
beta = (1 - savings) * concentration
print(f"Using beta distribution with alpha={alpha}, beta={beta} for savings={savings}")
```

```python
num_samples = 1000
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import beta as beta_dis
x = np.linspace(0, 1, num_samples)
y = beta_dis.cdf(x, alpha, beta)
plt.plot(x, y, label=f'Beta({alpha}, {beta})')
plt.title('Beta Distribution')
plt.xlabel('x')
plt.ylabel('Probability Density')
plt.legend()
plt.grid()
plt.show()
```
