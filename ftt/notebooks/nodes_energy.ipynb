```python
import pandas as pd
import os
from nnt.profiling.nvidia_profiler import NvidiaProfiler
pd.set_option("display.width", 1000)
```

```python
DIR = "/sc/projects/sci-herbrich/chair/lora-bp/vincent.eichhorn/nnt/nodes"
# create for energy.csv files in subdirectories
energy_files = []
for root, dirs, files in os.walk(DIR):
    for file in files:
        if file == "energy_log.csv":
            energy_files.append(os.path.join(root, file))
data = []
for f in energy_files:
    if "0-nlayer" in f or "1-nlayer" in f:
        continue
    node = f.split("/")[-5]
    print(f"Processing {f} for node {node}")
    prof = NvidiaProfiler.from_cache(f)
    energy = prof.get_total_energy(record_steps=["step_begin"])
    time = prof.get_total_time(record_steps=["step_begin"])
    data.append({
        "node": node,
        "energy": energy,
        "time": time,
    })
df = pd.DataFrame(data)
print(df)

print(df["energy"].describe())
print(df["time"].describe())
```
